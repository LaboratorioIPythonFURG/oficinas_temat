<h2><b>
Limites de funções
</b></h2>

O SymPy oferece a função <b>limit</b> para lidar com limites de funções simbólicas, a estrutura desse comando é:

<hr>
 <center><b>limit(</b> &lt;expressão&gt;<b>,</b>&lt;variável&gt;<b>,</b>&lt;a&gt;<b>,</b>&lt;lado&gt;<b>)</b></center>
<hr>

Onde <b>a</b> é a constante tal que $x \rightarrow a$ e <b>lado</b> representa o limite lateral à esquerda (<b><font color='#e65100'>'-'</font></b>) e à direita (<b><font color='#e65100'>'+'</font></b>).

<font color='#008000'><b>Exemplo (assíntotas): </b></font> Determine as assíntotas verticais e horizontais, se houver, da função $h(x)$ definida por:

$$ h(x) = \frac{3x}{\sqrt{2x^2-4x}} $$

&nbsp; <b>Passo 1</b>: Carregar a biblioteca.

```python
from sympy import  *
```

&nbsp; <b>Passo 2</b>: Declarar as variáveis.

```python
x = symbols('x')
```

&nbsp; <b>Passo 3</b>: Declarar a função $h(x)$.

```python
def h(x):
    return 3*x/sqrt(2*x**2-4*x)

h(x)
```

&nbsp; <b>Passo 4</b>: Utilizar comando <b>solve</b> para resolver $ 2x^2-4x = 0$ e determinar os valores de $x$ para os quais a função $h$ não está definida.

```python
pontos = solve(2*x**2-4*x,x)
pontos
```

```python
p1 = pontos[0]
p1
```

```python
p2 = pontos[1]
p2
```

&nbsp; <b>Passo 5</b>: Utilizar comando <b>limit</b> para determinar a assíntota vertical em <b>p1</b>.

```python
limit(h(x),x,p1,'-')
```

&nbsp; <b>Passo 6</b>: Utilizar comando <b>limit</b> para determinar a assíntota vertical em <b>p2</b>.

```python
limit(h(x),x,p2,'+')
```

&nbsp; <b>Passo 7</b>: Utilizar comando <b>limit</b> para determinar as assíntotas horizontais .

```python
h1 = limit(h(x),x,oo)
h1
```

```python
h2 = limit(h(x),x,-oo)
h2
```

&nbsp; <b>Passo 8</b>: Utilizar comando <b>plot</b> para exibir $h(x)$ .

```python
plot(h(x),h1,h2,legend=True,ylim=[-3,10])
```

<hr>
<font color="red"><b>ATENÇÃO:</b></font>

Note que SymPy calcula erroneamente o limite:

```python
limit(h(x),x,p1,"+")
```

Não faz sentido estudarmos este limite, pois os valores muito pequenos à direita de $0$ não pertencem ao domínio de $h$. Portanto, ao utilizar o SymPy para estudar funções com pontos de descontinuidade, é necessário verificar se o resultado faz sentido matematicamente.

<font color='#008000'><b>Exemplo (continuidade): </b></font>  O raio da Terra é de aproximadamente $6 \ 400 \ \textrm{km}$ quilômetros; um corpo situado a $x$ quilômetros do centro da Terra pesa $w(x) \ \textrm{kg}$, onde,

$$ \begin{cases}
 & Ax  \text{ para } x\leq 4 \ 000 \\
 & B/x^2 \text{ para } x >  4 \ 000
\end{cases} $$

e $A$ e $B$ são constantes positivas. Qual deve ser a relação entre $A$ e $B$ para que $w(x)$ seja contínua para qualquer valor de $x$?

&nbsp; <b>Passo 1</b>: Carregar a biblioteca.

```python
from sympy import *
```

&nbsp; <b>Passo 2</b>: Declarar as variáveis.

```python
A = Symbol('A',positive=True)
B = Symbol('B',positive=True)
x = Symbol('x')
x
```

&nbsp; <b>Passo 3</b>: Calcular limite lateral à esquerda:

```python
le = limit(A*x,x,4000,'-')
le
```

&nbsp; <b>Passo 4</b>: Calcular limite lateral à direita:

```python
ld = limit(B/x**2,x,4000,'+')
ld
```

&nbsp; <b>Passo 5</b>: Estabelecer condições para $A$ e $B$:

```python
expr = solve(le-ld,A)[0]
expr
```

&nbsp; <b>Passo 6</b>: Verificar condições de continuidade.

```python
le = limit(expr*x,x,4000,'-')
ld = limit(B/x**2,x,4000,'+')
[le,ld]
```

<p style="color:#404040"><font color="#233242"><b>Solução da Questão Norteadora</b></font>
    
 Queremos determinar a posição da massa ao ser abandonada depois de muito tempo sabenqdo que a posição do carrinho é dada por  $$ x_{t} = e^{-0,08t}[\  \cos(2t) - \mbox{sen}(2t) ]\ $$

&nbsp; <b>Passo 1</b>: Carregar a biblioteca.

```python
from sympy import *
```

&nbsp; <b>Passo 2</b>: Declarar as variáveis.

```python
t = symbols('t')
t
```

&nbsp; <b>Passo 3</b>: Declarar a função $x_{t}(t)$.

```python
def xt(t):
    return exp(-0.08*t)*(cos(2*t)-sin(2*t))

xt(t)
```

&nbsp; <b>Passo 3</b>: Utilizar a função <b>limit</b> para calcular $\displaystyle \lim_{ t \to \infty } x_{t}(t)$.

```python
limit(xt(t),t,oo)
```

&nbsp; <b>Passo 5</b>: Utilizar comando <b>plot</b> para exibir $x_{t}(x)$ .

```python
plot(xt(t),ylim=[-1.5,1.5],xlim=[0,10],ylabel='Posição',title='Posição da massa ao longo do tempo')
```
