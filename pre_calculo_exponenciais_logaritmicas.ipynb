## **3. Potenciação**

A potenciação é uma operação matemática que representa a multiplicação de um número por ele mesmo várias vezes. Dado um número real $a$ (base) e um número inteiro $n$ (expoente), temos:

$$a^n = \underbrace{a \times a \times \cdots \times a}_{n\text{ vezes}}$$

### **Propriedades da Potenciação**

1. **Base 1:** $1^n = 1$
2. **Expoente 0:** $a^0 = 1$ (com $a \neq 0$)
3. **Expoente 1:** $a^1 = a$
4. **Produto de potências de mesma base:** $a^m \cdot a^n = a^{m+n}$
5. **Quociente de potências de mesma base:** $\frac{a^m}{a^n} = a^{m-n}$
6. **Potência de potência:** $(a^m)^n = a^{m \cdot n}$
7. **Potência de produto:** $(ab)^n = a^n b^n$
8. **Potência de quociente:** $(\frac{a}{b})^n = \frac{a^n}{b^n}$

### **Exemplos Resolvidos de Potenciação**

## **4. Funções Exponenciais**

Uma função exponencial é uma função da forma $f(x) = a^x$, onde $a$ é um número real positivo diferente de $1$. O número $a$ é chamado de base da função exponencial.

### **Propriedades das Funções Exponenciais**

1. Se $a > 1$, a função é crescente.
2. Se $0 < a < 1$, a função é decrescente.
3. O domínio é $\mathbb{R}$ (todos os reais).
4. A imagem é $\mathbb{R}_+^*$ (reais positivos).
5. O gráfico sempre passa pelo ponto $(0,1)$.
6. Não possui raízes reais.

### **5.1 Crescimento Exponencial**

 A lei do crescimento exponencial pode ser escrita em formas diferentes, mas matematicamente equivalentes, usando uma base diferente . As formas mais comuns são as seguintes:

$$x(t)= x_0 \cdot (1+k)^t = x_0 \cdot e^{kt}$$

ou

$$
x(t)=x_0 \cdot 2^{\frac{t}{T}}
$$

onde:
- $x_0$ é a quantidade inicial;
- $t$ é a quantidade de tempo;
- $k$ é a taxa de crescimento;
- $T$ é o tempo que leva para duplicar.


**Obs**: Parâmetros negativos para decaimento.

### **Exemplos Resolvidos de Crescimento Exponencial**

<span style="color: red;"><b>Exemplo 1</b></span>:  Determinada população de bactérias cresce exponencialmente, com o tempo, com população inicial de $40000$  e uma taxa de crescimento de $0,2$.


biology lab GIF by University of California.gif

a) Qual será o número aproximado de bactérias após 5 minutos? Use a equação:

$$
P(t)= P_0\cdot e^{kt}
$$

onde $t$ é o tempo em minutos.

```python
P0 = 40000  # Quantidade inicial de bactérias
k = 0.2   # Taxa de crescimento
t = 5     # Tempo em minutos

P = P0 * np.exp(k*t)

print(f'Quantidade de bactérias após {t} horas: {P:.2f}')
```

b) Qual é o número inicial de bactérias?

```python
P0 = 40000  # Quantidade inicial de bactérias
r = 0.2   # Taxa de crescimento
t = 0     # Tempo em horas

P = P0 * np.exp(r*t)

print(f'Quantidade de bactérias após {t} horas: {P:.2f}')
```

<span style="color: red;"><b>Exemplo 2</b></span>: Imagine um algoritmo de busca que, devido à sua construção ineficiente, exibe um comportamento de crescimento exponencial. Para um tamanho de entrada $n = 1$, o algoritmo leva $0,5$ milissegundos (ms) para ser executado. Determine o tempo de execução do algoritmo para uma entrada de tamanho $n = 5$, sabendo que

$$T(n) = T_0 \cdot 2^{n}.$$

```python
T0 = 0.25  # Tempo incial da primeira execução
n = 5     # Tamanho da entrada de dados

T = T0 * 2**n

print(f'Tempo de execução para {n} dados: {T:.2f}')
```

### **5.2 Decaimento Exponencial**

O decaimento exponencial é o processo oposto, onde a quantidade diminui a uma taxa proporcional ao seu valor atual.

### **Exemplos Resolvidos de Decaimento Exponencial**

<span style="color: red;"><b>Exemplo 1</b></span>: Uma substância radioativa tem meia-vida de $2$ horas. Se inicialmente há $100$g, quanto restará após $6$ horas? Use a função:

$$
S(t)= S_0.2^{\frac{-t}{T}}
$$

```python
S0 = 100  # Quantidade inicial da substância
T = 2   # Tempo que leva para dividir por dois a substância
t = 6     # Tempo em horas

S = S0 * 2**(-t/T)

print(f'Quantidade de substância após {t} horas: {S:.2f}')
```

```python
import matplotlib.pyplot as plt
import numpy as np

S0 = 100  # Quantidade inicial da substância
T = 2   # Tempo que leva para dividir por dois a substância

# Define o intervalo de tempo para o gráfico
t_values = np.linspace(0, 10, 100) # Gera 100 pontos entre 0 e 10 horas

# Calcula a quantidade de substância para cada valor de tempo
S_values = S0 * 2**(-t_values / T)

# Plota o gráfico
plt.figure(figsize=(10, 6))
plt.plot(t_values, S_values)
plt.xlabel('Tempo (horas)')
plt.ylabel('Quantidade de substância')
plt.title('Decaimento da Substância')
plt.grid(True)
plt.show()
```

<span style="color: red;"><b>Exemplo 2</b></span>: Um medicamento é eliminado do organismo a uma taxa de $10\%$ por hora. Se a dose inicial é $200mg$, quanto restará após $3$ horas? Use:

$$R(t) = R_0 \cdot (1-r)^{t}$$

```python
R0 = 200  # Quantidade inicial de medicamento
r = 0.1   # Taxa de eliminação
t = 2     # Tempo em horas

R = R0 * (1 - r)**t

print(f'Quantidade de medicamento após {t} horas: {R:.2f}')
```

$$

### **7.1 Escalas Logarítmicas**

As escalas logarítmicas são muito utilizadas em diversas áreas da ciência e engenharia. Alguns exemplos:

1. **Escala Richter** para medir a magnitude de terremotos.
2. **Escala de Decibéis** para medir a intensidade do som.
3. **Escala de pH** para medir a acidez de soluções.

### **Exemplos Resolvidos de Escalas Logarítmicas**

<span style="color: red;"><b>Exemplo 1</b></span>: Calcule a magnitude de um terremoto com amplitude $1000 mm$ com tempo desde o início do trem de ondas P (primárias) até à chegada das ondas S (secundárias) de $30s$. Use a fórmula:

$$
M=log_{10}(\frac{A \cdot \Delta t^3}{1,62})
$$

onde:

- $A$: amplitude das ondas sísmicas, em milímetros, medida directamente no sismograma;
- $\Delta t$:  tempo, em segundos, desde o início do trem de ondas P (primárias) até à chegada das ondas S (secundárias);
- $M$: magnitude arbitrária, mas constante, aplicável a sismos que libertem a mesma quantidade de energia.

```python
A = 1000     # Amplitude em mm
Dt = 30   # Variação de tempo em s

M = np.log10(((A*Dt**3)/1.62))

print(f'Magnitude do terremoto na escala Richter: {M:.2f}')
```

<span style="color: red;"><b>Exemplo 2</b></span>: Considere uma onda sonora de intensidade igual a $9,8 \cdot 10^{-5} W/m^2$ calcule seus decibels. Use a fórmula:

$$
dB = 10 \cdot log_{10}(\frac{I}{I_0})
$$

onde:

- $I_0$: limiar da audibilidade ($10^{-12} W/m^2$);
- $I$: intensidade sonora observada.


```python
I0 = 1e-12     # Intensidade de referência em W/m^2
I = 9.8e-5   # Intensidade medida em W/m^2

dB = 10 * np.log10(I/I0)

print(f'Intensidade sonora em decibels: {dB:.2f}')
```

<span style="color: red;"><b>Exemplo 3</b></span>: Dada uma solução com valor da concentração do hidrônio $[H^+]=2 \cdot 10^{-7}$ calcule o pH da solução. Use:

$$
pH = - log_{10}([H^+])
$$

onde:

- $[H^+]$: valor da concentração do hidrônio.

```python
import numpy as np
import matplotlib.pyplot as plt

# Cria um array de valores de pH de 0 a 14
ph_values = np.linspace(0, 14, 100)

# Calcula a concentração de hidrônio para cada valor de pH
# Usando a relação [H3O+] = 10^-pH
hydronium_concentration = 10**(-ph_values)

# Cria o gráfico
plt.figure(figsize=(10, 6))
plt.plot(ph_values, hydronium_concentration)

# Configura os rótulos e título do gráfico
plt.xlabel('pH')
plt.ylabel('Concentração de Íons Hidrônio ([H3O+])')
plt.title('Concentração de Íons Hidrônio em Função do pH')


# Adiciona uma grade ao gráfico
plt.grid(True, which="both", linestyle='--')

# Exibe o gráfico
plt.show()
```

```python
H = 2.1e-7     # Concentração em mol/L

pH = -np.log10(H)

print(f'pH da solução: {pH:.2f}')
```

### **7.2 Crescimento Logarítmico**

O crescimento logarítmico é um processo onde a taxa de crescimento diminui com o tempo. É modelado pela função:

$$P(t) = P_0 \cdot \ln(t + 1)$$

onde:
- $P_0$ é a quantidade inicial
- $t$ é o tempo

### **Exemplos Resolvidos de Crescimento Logarítmico**

```python
h = 100

T = 10 * np.log(h + 1)

print(f'Tempo de aprendizado: {T:.2f}')
```

<span style="color: red;"><b>Exemplo 2</b></span>: O crescimento de uma árvore é modelado por $A(t) = 2 \cdot \ln(t + 1)$, onde $t$ são os anos. Qual a altura após $5$ anos?



```python
t = 5

A = 2 * np.log(t + 1)

print(f'Altura da árvore: {A:.2f}')
```

<span style="color: red;"><b>Exemplo 3</b></span>: A produtividade de um trabalhador é modelada por $P(t) = 5 \cdot \ln(t + 1)$, onde $t$ são os meses de experiência. Qual a produtividade após $3$ meses?

<span style="color: green;"><b>Exemplo 2:</b></span> Calcule o valor de um ativo que se deprecia 10% ao ano, sabendo que seu valor inicial era R$ 5000,00 e já se passaram 2 anos.

```python
# Resolução do Exemplo 2
V0 = 5000  # Valor inicial
d = 0.10   # Taxa de depreciação
t = 2      # Tempo em anos

V = V0 * (1 - d) ** t
print(f'Valor após {t} anos: R$ {V:.2f}')
```
