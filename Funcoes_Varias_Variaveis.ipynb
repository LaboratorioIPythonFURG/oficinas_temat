<h2 style='color:#0000ff'>6 &nbsp; Esboço do domínio de funções de várias variáveis com o plot_implicit</h2>

&nbsp; A função SymPy <b>plot_implicit</b> é utilizada para exibir regiões ou curvas de equações/inequações. Essa funcionalidade pode auxiliar a reconhecer a região onde os pares $(x,y)$ estão restritos. Vamos considerar a curva $z=f(x,y)$ no $\mathbb{R}^3$:

$$f(x,y)=\sqrt[]{1-x^2-y^2}$$

&nbsp; O domínio $D(f)$ da função é dado por:

$$ D(f) : 1-x^2-y^2 \geq 0$$

&nbsp; Para visualizar a região que satisfaz a restrição $D(f)$, basta utilizar a função <b>plot_implicit</b> da seguinte forma:

```python
plot_implicit(1-x**2-y**2 >= 0, (x,-2,2),(y,-2,2))
```

<h4>Obtendo curvas de nível com as bibliotecas <b>NumPy</b> e <b>Matplotlib</b> </h4>

&nbsp; <b>Passo 1</b> Carregas as bibliotecas <b>NumPy</b> e <b>Matplotlib</b>:

```python
import numpy as np
import matplotlib.pyplot as plt
```

&nbsp; <b>Passo 2</b> Gerar vetores $x$ e $y$:

```python
vx = np.arange(-10,50,0.1)
vy = np.arange(-25,25,0.1)
```

&nbsp; <b>Passo 3</b> Criando a malha $vx \times vy$:

```python
vx, vy = np.meshgrid(vx, vy)
```

&nbsp; <b>Passo 4</b> Obtendo a superfície $V$ sobre a malha $vx \times vy$:

```python
V =  900/np.sqrt(vx**2+vy**2)+3600/np.sqrt((vx-30)**2+vy**2)
```

&nbsp; <b>Passo 5</b> criando vetor contendo os níveis:

```python
niveis = [100,250,500,1000,1500]
```

&nbsp; <b>Passo 6</b> Exibir a figura:

```python
#cria subplots
fig, ax = plt.subplots()

#comando 'contour'
cs = ax.contour(vx, vy, V,niveis, cmap=plt.cm.YlGnBu_r)

#Propriedades de estilo
ax.clabel(cs, inline=0.5, fontsize=10)
ax.grid(linestyle='-')
ax.set_title('Curvas equipotenciais')
```

&nbsp; <b>Passo 7</b> Exibir figura 3d:

```python
import matplotlib.pyplot as plt
import numpy as np

fig = plt.figure()
ax = fig.add_subplot(projection='3d')

#Plotar a superfície
ax.plot_surface(vx, vy, V, cmap=plt.cm.YlGnBu_r)

# Inserir os limites e nomear eixos
ax.set_zlim(100, 3000)
ax.set_xlabel(r'$x$')
ax.set_ylabel(r'$y$')
ax.set_zlabel(r'$V$')

#Exibir figura
plt.show()
```
