## **3. Trigonometria**



$\quad$ A trigonometria é uma parte da matemática que estuda as relações entre os ângulos e os lados dos triângulos. Ela é fundamental no cálculo de distâncias, alturas e movimentos. Além disso, é utilizada no estudo de movimentos periódicos, como ondas e rotações.

### **Teorema de Pitágoras**

$\quad$ O Teorema de Pitágoras estabelece que, em um triângulo retângulo, a soma dos quadrados das medidas dos catetos é igual ao quadrado da medida da hipotenusa. A relação matemática é dada por:

$$
\text{a}^2 = \text{b}^2 + \text{c}^2.
$$

$\quad$ Um corolário importante desse teorema é que, quando os dois catetos possuem o mesmo comprimento, a hipotenusa é igual ao valor do cateto multiplicado por $\sqrt{2}$. Ou seja, se $\text{b} = \text{c}$, então:

$$
\text{a} = \text{b} \sqrt{2}.
$$


**EXEMPLO:**

### **Seno**

$\quad$ O seno de um ângulo é a razão entre o cateto oposto e a hipotenusa.

$$
\mbox{sen}(\theta) = \frac{\text{cateto oposto}}{\text{hipotenusa}}
$$



### **Cosseno**

$\quad$ O cosseno de um ângulo é a razão entre o cateto adjacente e a hipotenusa.

$$
\cos(\theta) = \frac{\text{cateto adjacente}}{\text{hipotenusa}}
$$

### **Tangente**

$\quad$ A tangente de um ângulo é a razão entre o cateto oposto e o cateto adjacente.

$$
\tan(\theta) = \frac{\text{cateto oposto}}{\text{cateto adjacente}} \longrightarrow  \tan(\theta) = \frac{\mbox{sen}(\theta)}{\cos(\theta)}
$$

### **3.1 Círculo Trigonométrico**

$\quad$ É um círculo com raio 1, desenhado num plano bidimencional com dois eixos (x,y), onde o centro está no ponto (0, 0). Ele é usado para mostrar como funcionam as funções seno, cosseno e tangente. Quando desenhamos um ângulo nesse círculo, o ponto em que ele termina tem coordenadas que mostram o valor do cosseno (no eixo x) e do seno (no eixo y) desse ângulo. O círculo ajuda a entender como esses valores mudam quando o ângulo aumenta, mesmo para ângulos maiores que 90° ou negativos. Também é muito útil para visualizar o comportamento das funções trigonométricas.

### **Tabela de Razões Trigonométricas**

$$
\begin{array}{|c|c|c|c|}
\hline
\theta & \mbox{sen}(\theta) & \cos(\theta) & \tan(\theta) \\
\hline
0^\circ & 0 & 1 & 0 \\
\hline

\hline

90^\circ & 1 & 0 & \text{indefinido} \\

\end{array}
$$

### **3.2. Relações Trigonométricas**

### **Identidade Fundamental da Trigonometria**

$\quad$ A mais importante das relações fundamentais é:

$$
\mbox{sen}^2(\theta) + \cos^2(\theta) = 1
$$

$\quad$ Essa identidade vale para qualquer valor real de $\theta$. Ela vem diretamente do círculo trigonométrico, onde pegando um ponto $\text{P} = (\cos(\theta),\mbox{sen}(\theta))$ pertence à circunferência de raio 1:

$$
\text{x}^2 + \text{y}^2 = 1 \Longrightarrow \cos^2(\theta) + \mbox{sen}^2(\theta) = 1
$$

### **Cossecante**

$\quad$ Dividindo ambos os lados por $\cos^2(\theta)$, obtemos:

$$
\frac{\mbox{sen}^2(\theta)}{\cos^2(\theta)} + \frac{\cos^2(\theta)}{\cos^2(\theta)} = \frac{1}{\cos^2(\theta)} \Longrightarrow \tan^2(\theta) + 1 = \sec^2(\theta)
$$

$\quad$ Sendo assim então:

$$
\sec(\theta) = \frac{1}{\cos(\theta)}
$$

### **Secante**

$\quad$ Dividindo ambos os lados por $\mbox{sen}^2(\theta)$, obtemos:

$$
\frac{\sin^2(\theta)}{\mbox{sen}^2(\theta)} + \frac{\cos^2(\theta)}{\mbox{sen}^2(\theta)} = \frac{1}{\mbox{sen}^2(\theta)} \Longrightarrow 1 + \cot^2(\theta) = \csc^2(\theta)
$$

$\quad$ Sendo assim então:

$$
\csc(\theta) = \frac{1}{\mbox{sen}(\theta)}
$$

### **Cotangente**

$\quad$ Como visto na demonstração da secante obtemos:

$$
\cot(\theta) = \frac{\cos(\theta)}{\mbox{sen}(\theta)} \longrightarrow \cot(\theta) = \frac{1}{\tan(\theta)}
$$



$\quad$ Implementando no Python essas relações Trigonométricas, temos:

````python
angulo = np.radians(139)

seno = np.sin(angulo)
cosseno = np.cos(angulo)
tangente = np.tan(angulo)
secante = 1/np.cos(angulo)
cossecante = 1/np.sin(angulo)
cotangente = 1/np.tan(angulo)

print(f'seno: {seno}')
print(f'cosseno: {cosseno}')
print(f'tangente: {tangente}')
print(f'secante: {secante}')
print(f'cossecante: {cossecante}')
print(f'cotangente: {cotangente}')
````

### **3.4. Operações com arcos**


#### **3.4.1 Adição**

* **Seno da soma**

$$
\mbox{sen}(a+b) = \mbox{sen}(a)\cos(b) + \cos(a)\mbox{sen}(b)
$$

* **Coseno da soma**

$$
\cos(a+b) = \cos(a)\cos(b) - \mbox{sen}(a)\mbox{sen}(b)
$$

#### **3.4.2 Subtração**

* **Seno da diferença**

$$
\mbox{sen}(a-b) = \mbox{sen}(a)\cos(b) - \cos(a)\mbox{sen}(b)
$$

* **Coseno da diferença**

$$
\cos(a-b) = \cos(a)\cos(b) + \mbox{sen}(a)\mbox{sen}(b)
$$

#### **3.4.3 Arco Dobro**

$\quad$ Existem quatro principais relações trigonométricas com arco dobro:

* $\mbox{sen}(2\theta)$:
$$
\mbox{sen}(2\theta) = 2\mbox{sen}(\theta)\cos(\theta)
$$

* $\cos(2\theta)$:
$$
\cos(2\theta) = \cos^2(\theta)-\mbox{sen}^2(\theta)
$$

* $\cos^2(\theta)$:
$$
\cos^2(\theta) = \frac{1+\cos(2\theta)}{2}
$$

* $\sin^2(\theta)$:
$$
\mbox{sen}^2(\theta) = \frac{1-\cos(2\theta)}{2}
$$

````python
import numpy as np
import matplotlib.pyplot as plt

def letra_G(t):
    x = np.cos(t) - 0.5*np.cos(2*t)
    y = np.sin(t) - 0.3*np.sin(2*t)
    x[t > np.pi] += 0.5  # abertura do G
    return x, y

def letra_L(t):
    x = 0.2 * np.cos(t)
    y = -t/np.pi
    return x, y

def letra_A(t):
    x = 0.5 * np.sin(t)
    y = np.abs(np.cos(t))
    return x, y

def letra_I(t):
    x = np.zeros_like(t)
    y = np.linspace(-1, 1, len(t))
    return x, y

def letra_S(t):
    x = np.sin(t)
    y = np.sin(t + np.pi/2)
    return x, y

def letra_E(t):
    x = np.cos(t)
    y = np.sin(2*t)/2
    return x, y

# Criação do gráfico
plt.figure(figsize=(12, 4))
plt.axis('equal')
plt.axis('off')

t = np.linspace(0, 2*np.pi, 300)

# Espaçamento horizontal entre letras
espacamentos = [0, 2, 4, 5, 6.5, 8]

funcoes_letras = [letra_G, letra_L, letra_A, letra_I, letra_S, letra_E]

for i, func in enumerate(funcoes_letras):
    x, y = func(t)
    x += espacamentos[i]
    plt.plot(x, y, color='darkred')

plt.title('Glaise', fontsize=20, style='italic')
plt.show()
````

````python
import numpy as np
import matplotlib.pyplot as plt

# Cada letra é feita com base em curvas senoidais estilizadas
def letra_G(t):
    x = 1.5*np.cos(t) - 0.3*np.cos(2*t)
    y = 1.5*np.sin(t) - 0.1*np.sin(3*t)
    x[t > np.pi] += 0.5
    return x, y

def letra_L(t):
    x = 0.1*np.sin(t)
    y = -t/np.pi
    return x, y

def letra_A(t):
    x = 0.5*np.sin(t)
    y = np.abs(np.cos(t))
    return x, y

def letra_I(t):
    x = np.zeros_like(t)
    y = np.linspace(-1, 1, len(t))
    return x, y

def letra_S(t):
    x = np.sin(t)
    y = np.sin(t + np.pi/2)
    return x, y

def letra_E(t):
    x = np.cos(t) * (1 - 0.2*np.sin(3*t))
    y = np.sin(2*t)/2
    return x, y

# Lista das funções e seus deslocamentos horizontais
funcoes_letras = [letra_G, letra_L, letra_A, letra_I, letra_S, letra_E]
espacamentos = [0, 2.2, 3.7, 5.0, 6.5, 8.2]

# Criação do gráfico
plt.figure(figsize=(12, 4))
plt.axis('equal')
plt.axis('off')

t = np.linspace(0, 2*np.pi, 400)

# Traçando cada letra
for i, func in enumerate(funcoes_letras):
    x, y = func(t)
    x += espacamentos[i]
    plt.plot(x, y, color='crimson', linewidth=2)

plt.title('Glaise', fontsize=24, style='italic')
plt.show()
````
