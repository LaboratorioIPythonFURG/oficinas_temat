<h2><b>3. Fundamentos de Equações Diferenciais</b>

<h3><b>3. 1 Introdução
    </b>
</h3>
Equações diferenciais são ferramentas matemáticas essenciais para modelar e resolver uma ampla variedade de problemas em diversos campos, como física, engenharia, biologia, economia e finanças. Elas descrevem a relação entre uma função e suas derivadas, capturando a ideia de como a taxa de mudança de uma quantidade está relacionada à própria quantidade.





my_animation.gif


<h3><b>3. 2 Definição
    </b>
</h3>

Uma **equação diferencial** é uma equação matemática que envolve uma ou mais derivadas de uma função desconhecida em relação a uma ou mais variáveis independentes. Essencialmente, ela expressa uma relação entre a função e sua taxa de variação. Uma **equação diferencial ordinária (EDO)** é uma equação que envolve uma
função desconhecida, $y(x)$, suas derivadas até uma ordem $n$ e a variável independente
$x$; ou seja, é uma equação da forma:

<center>
$
f(x,y,y',y'',...,y^{(n)})=0.
$




<h3><b>3. 3 Solução geral
    </b>
</h3>

Uma equação diferencial ordinária geralmente possui um número inﬁnito
de soluções. Chama-se família de soluções o conjunto formado por todas as soluções
da EDO. Em problemas de modelagem o gráﬁco da família de soluções tem grande
importância, pois mostra o formato das soluções. Geometricamente, a solução geral de uma equação diferencial,
é representada por inﬁnitas curvas, uma para cada valor das constantes arbitrárias.
Estas curvas representam uma família de curvas que recebem o nome de curvas
integrais.

<h3><b>3. 4 Problemas de valor inicial
    </b>
</h3>

Um problema de valor inicial (PVI) consiste em uma equação diferencial
sujeita a determinadas condições pré-estabelecidas, ou seja, condições subsidiárias
relativas à função incógnita e suas derivadas tudo dado para um mesmo valor da
variável independente. Tais condições são ditas condições iniciais.
Um problema de valor inicial (PVI) de primeira ordem consiste em
resolver a equação diferencial

<center>
$$
\frac{dy}{dx} = f(x,y)
$$
</center>

sujeita à condição inicial

<center>
$$
y(x_0) = y_0
$$
</center>

onde $x_0$ é um número no intervalo $I$ e $y_0$ é um número real arbitrário.

<h3><b>3. 5 Problemas de valor de contorno
    </b>
</h3>

Um problema de valor de contorno (PVC) consiste em resolver uma
equação diferencial de ordem $2$ ou maior na qual a variável dependente $(y)$ ou suas derivadas são especiﬁcadas em pontos diferentes.

O problema
<center>
$$
a_2\frac{d^2y}{dx^2}+a_1\frac{dy}{dx} + a_0y=g(x)
$$
</center>

sujeito a

<center>
$$
y(a)=y_0, y(b) = y_1
$$
</center>

é chamado de problema de valor de contorno. Os valores $y(a) = y_0$ e $y(b) = y_1$ são
chamados de condições de contorno ou condições de fronteira.
Dessa forma, uma solução para um PVC é uma função $y = y(x)$ que
satisfaz não só à equação diferencial dada, mas também a todas as condições de
contorno.

<h3><b>4. 3 Problemas de valor de contorno</b></h3>

Para problema de valor de contorno define-se as condições de contorno. Após isso, define-se a solução geral e substitui suas variáveis de acordo com as condições. Assim, tem-se um sistema que é  solucionado para obter os valores das constantes


<font color='#008000'><b>&nbsp; Exemplo  </b></font>Seja $N(t$) a quantidade de uma substância ou população sujeita a um processo de crescimento ou decrescimento (decaimento). Se admitirmos que $\frac{dN}{dt}$, taxa de variação da quantidade de substância ou população, é proporcional à quan-
tidade de substância presente, tem-se a equação diferencial,

<center>
$$
\frac{dN}{dt}=kN,
$$
</center>

onde $k$ é a constante de proporcionalidade. Diz-se que $k$ é uma constante de crescimento, se $k > 0$ ou de decrescimento (decaimento), se $k < 0$. E sabendo que uma cultura de bactérias cresce a uma taxa proporcional à quantidade presente. Após uma hora, observamos $1.000$ núcleos de bactérias na cultura, e após $4$ horas, $3.000$ núcleos. Determine uma expressão para o número de núcleos presentes na cultura no tempo arbitrário
$t$;

&nbsp; <b>Passo 1</b>: Declarar as variáveis.

````python
t = symbols('t')
k = symbols('k', real=True)
C1 = symbols('C1')
N = Function('N')
````

&nbsp; <b>Passo 2</b>: Declarar a EDO que modela o problema.



````python
edo = Derivative(N(t),t)-k*N(t)
````

&nbsp; <b>Passo 3</b>: Resolver a solução geral.


````python
sol_ger = dsolve(edo, N(t))
sol_ger
````

&nbsp; <b>Passo 4</b>: Substituir as condições de contorno na solução geral. Utilizando a função *.rhs* para atribuir o lado direito da equação.

````python
cond1 =  sol_ger.rhs.subs(t ,1)-1000
cond2 =  sol_ger.rhs.subs(t ,4)-3000
````

&nbsp; <b>Passo 5</b>: Solucionar um sistemas com essas duas equações para descobrir os valores das constantes e a partir delas construir a solução particular.

````python
constantes = solve((cond1, cond2), (C1, k))[0]

sol_par = sol_ger.subs(C1, constantes[0]).subs(k, constantes[1])
sol_par
````

Uma massa de $0,25kg$ é atada a uma mola com constante de
elasticidade igual a $4$N/cm. Supondo que uma força de amortecimento igual ao
dobro da velocidade instantânea atue no sistema, determine a equação do movimento
se o peso parte da posição de equilíbrio com velocidade de $3$cm/s para cima.




m2-res_1080p.gif

Segundo a Lei de Newton, a equação diferencial do movimento é,

<center>
$$
m\frac{d^2x}{dt^2}=-\beta \frac{dx}{dt}-kx
$$
</center>

Para este exemplo tem-se $k=4$, $\beta=2$ e $m=0,25$. Sabendo que a posição inicial é $0$ então $x(0)=0$ e sabendo que a velocidade inicial é $3$ então $x'(0)=-3$, assim, define-se as condições de contorno.

&nbsp; <b>Passo 1</b>: Declarar a variáveis e constantes.

````python
t = Symbol("t")
x = Function("x")

k=4
beta=2
m=2
````

&nbsp; <b>Passo 2</b>: Declarar a EDO que modela o problema.



````python
edo = m*Derivative(x(t),t,2)+beta*Derivative(x(t),t)+k*x(t)
edo
````

&nbsp; <b>Passo 3</b>: Resolver a solução geral.


````python
sol_ger = dsolve(edo, x(t))
sol_ger
````

&nbsp; <b>Passo 4</b>: Substituir as condições de contorno na solução geral. Utilizando a função *.rhs* para atribuir o lado direito da equação.

````python
cond1 =  sol_ger.rhs.subs(t ,0)
cond2 =  (Derivative(sol_ger.rhs, t).doit()).subs(t, 0)+3
cond2
````

&nbsp; <b>Passo 5</b>: Solucionar um sistemas com essas duas equações para descobrir os valores das constantes e a partir delas construir a solução particular.

````python
C1 = symbols('C1')
C2 = symbols('C2')

constantes = solve((cond1, cond2), (C1, C2))

sol_par = sol_ger.subs(C1, constantes[C1]).subs(C2, constantes[C2])
sol_par
````

````python
plot(sol_par.rhs, (t, 0, 20))
````
